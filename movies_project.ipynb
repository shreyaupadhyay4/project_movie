```python
# Q3. How do we clean the `duration` column? (e.g., `'175 min'` → `175`)
```

```python
df["duration"]
```

```python
# Q4. How do we clean the `gross` column and `votes` column?
```

```python
# Q22. Which Director + Actor duo is the most frequent?
```

```python
# Step 1: Explode directors if it's  multiple
a = df[['directors_name', 'stars_name']].explode('directors_name')


```

```python

# Step 2: Group by director + actor and count occurrences
b = a.groupby(['directors_name', 'stars_name']).size().reset_index(name='count')
```

```python

# Step 3: Sort by frequency and get the top duo
b.sort_index( ascending=False).head(1)
```

```python
# Q24. What is the best genre for a debut director? (Highest avg rating for 1st time directors)
```

```python

# Step 1: Count how many films each director has
director_counts = df.groupby('directors_name')['directors_name'].size().reset_index(name='film_count')


```

```python
# Step 3: Filter only debut directors (film_count == 1)
debut_directors = director_counts[director_counts['film_count'] == 1]


```

```python
resultt = debut_directors.groupby('genre')['rating'].mean().sort_values( ascending=False).head(1)


```
